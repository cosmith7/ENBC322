**Problem 1:**
Merge two sorted arrays.

    Input: Two sorted arrays
    
    Output: A merged sorted array

```python
'''
merge_sorted_arrays(arr1, arr2)
approach: compare each beginning element of the sorted arrays. add the smaller one to the resulting array
'''
def merge_sorted_arrays(arr1, arr2):
    result = []
    i = 0
    j = 0

    while i < len(arr1) and j < len(arr2):
        arr1_min = arr1[i]
        arr2_min = arr2[j]
        
        # check if the first array's element is smaller than the second's
        if arr1_min <= arr2_min:
            print(f"{arr1_min} is less than {arr2_min}")
            result.append(arr1_min) # add to the resulting array
            i += 1
        
        # otherwise, add the second element instead
        else:
            print(f"{arr1_min} is more than {arr2_min}")
            result.append(arr2_min)
            j += 1

        print(f"Current sorted array: {result}")
        
    # at the end of the while loop, one of the arrays will be empty, while the other has remaining elements
    # these remaining elements can be safely added to the resulting sorted array
    result += arr1[i:]
    result += arr2[j:]

    return result

# examples
array_one = [-1, 3, 5, 7]
array_two = [2, 6, 7, 11]

sorted_merge = merge_sorted_arrays(array_one, array_two)
print(sorted_merge)
            
```

**Problem 2:** Implement merge sort without recursion.

```python
'''
merge_sort_iteratively(arr)
approach: we divide the array non-recursively into elements with length = 1, then use the merge_sorted_arrays(arr1, arr2) function from part one
'''

def merge_sort_iteratively(arr):

    # keep track of the sorted arrays' length using width
    width = 1

    while width < len(arr):
        merged = []

        # deal with two sorted arrays at a time
        for i in range(0, len(arr), width * 2):

            # get the first sorted arr
            left = arr[i:i + width] # width of the arr increases on each iteration

            # get the second sorted arr
            right = arr[i + width: i + width * 2]

            # sort the two 
            merged += merge_sorted_arrays(left, right)

        arr = merged
        width *= 2

    return arr

# example
my_array = [9, -1, 4, 2, 6, 1, 5, 3, 10]
print(merge_sort_iteratively(my_array))
```

**Problem 3:** Implement merge sort using recusion.

```python
'''
merge_sort_recursion(arr)
approach: we divide the array recursively into elements with length = 1, then use the merge_sorted_arrays(arr1, arr2) function from part one
'''

def merge_sort_recursion(arr):
    # base case:
    if len(arr) <= 1:
        return arr

    # 1. split our arr into a left and right half
    middle = len(arr) // 2
    left = arr[:middle]
    right = arr[middle:]
    
    # 2. recursively split again by calling the function of the left and right
    left = merge_sort_recursion(left)
    right = merge_sort_recursion(right)

    # 3. use Problem 1's merge_sorted_arrays() to combine
    return merge_sorted_arrays(left, right)

# example
my_array = [9, -1, 4, 2, 6, 1, 5, 3, 10]
print(merge_sort_recursion(my_array))
```
