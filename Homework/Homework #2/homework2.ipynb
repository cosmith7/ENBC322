**Problem 1:**

Given an array of integers, nums, where each integer is between 1 and n
(n is the size of the array), some elements appear twice and others appear
once. Return an array of all the integers that appear twice in the array. You
must return the integers in their original order.

Example:

Input: [4, 3, 2, 7, 8, 2, 3, 1]

Output: [2, 3]

````python
'''
Problem 1
'''

input = [4, 3, 2, 7, 8, 2, 3, 1]

def find_duplicates(arr):
    result = [] # array to store the duplicates
    searched = [] # array for input navigation, so the duplicate integers are returned in their original order
    
    for i in arr: # search the input
        searched.append(i) # add the input element to the searching array
        if searched.count(i) > 1: # use count() to determine if a element occurred more than once
            result.append(i) # if it occurred more than once, add to results

    print(result)

find_duplicates(input)
````

**Problem 2:**

Review Merge Sort. Write a Python function that implements Merge Sort and provide an example demonstrating how it works.

*See the print statements for my example.*


````python
# merge sort algorithm

input_even = [4, 3, 2, 7, 8, 2, 3, 1]
input_odd = [5, 2, 1, 5, 19, 3, -2, 18, 4]

def merge_sort(arr):
    # base case
    if len(arr) <= 1:
        return arr

    # 1) divide the array into parts
    middle = len(arr) // 2
    half_1 = arr[:middle]
    half_2 = arr[middle:]
    print("Dividing array into", half_1, "and", half_2)

    # 2) recursively sort each half
    left = merge_sort(half_1)
    right = merge_sort(half_2)
    print("\nFirst to merge: ", left)
    print("Second to merge: ", right)

    # use helper function to merge the two halves
    return merge_sort_helper(left, right)

# helper function for sorting and merging two arrays
def merge_sort_helper(arr1, arr2):
    result = []
    
    while len(arr1) > 0 and len(arr2) > 0:
        if arr1[0] < arr2[0]: # check if the first elem of the left array is less
            print("Adding", arr1[0], "to the array.")
            result.append(arr1[0]) # add the first elem of the left array first
            arr1 = arr1[1:] # remove that elem from the left array
        else:
            print("Adding", arr2[0], "to the array.")
            result.append(arr2[0]) # add the elem of the right array instead, since it is less/equal
            arr2 = arr2[1:] # remove that elem from the right array
        
    # add anything that remains, since it must be greater than the added elements
    result += arr1
    result += arr2 
    print("Sorted array:", result, "\n")

    return result

print("Example one: Even array. Starting array: ", input_even, "\n")
merge_sort(input_even)

print("\n----------------------------\n")

print("Example two: Odd array. Starting array: ", input_odd, "\n")
merge_sort(input_odd)

````

**Problem 3:**

Study Insertion Sort and provide an analysis of its time complexity.

**Insertion Sort** works by separating an array into a sorted and unsorted half, then *inserting* the unsorted elements into the sorted list appropriately. Looking at the following example:

``` {python}
arr = [8, 5, 3, 6, 1, 4, 2, 10]
```

Lets say we have a list size of N. We start with the first element (8) as our sorted half, and classify the remaining elements as our unsorted half. We take the first element of the unsorted list (5), and *insert* it into the sorted half appropriately.

``` {python}
arr = [5, 8, 3, 6, 1, 4, 2, 10]
```

5 was *inserted* before 8 since it is less. We continue the sort with the next element of the unsorted half, 3.

``` {python}
arr = [3, 5, 8, 6, 1, 4, 2, 10]
```

We continue this process until the list is sorted.

``` {python}
arr = [1, 2, 3, 4, 5, 6, 8, 10]
```

We go through each element of the list, which takes *O(N)* iterations. We then take each element and sort it within our sub-list, which can take up to *O(N)* iterations due to comparing and shifting elements. Thus, our resulting time complexity is the product of the two, **O(N^2)**.
